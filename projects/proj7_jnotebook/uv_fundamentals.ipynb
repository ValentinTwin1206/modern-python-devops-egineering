#### 1.1.2 Virtual Environments

So far we have used ``uv`` to setup an initial project for us. Normally if you are working on a ``Python`` based implementation the standard way is to work with a so called **virtual environment**. Without ``uv`` this becomes quite messy, because YOU have to take care about the creation, deletion and update of the environment according to your project. Thanks to ``uv`` we can faciliate the certain commands to integrate and manage **virtual environments** with our project. 

> NOTE: virtual environments are created and updated throughout multiple uv commands - we'll refer to this during the next chapters

For now we are going to use the ``uv venv`` command, which creates a virtual environment as described at the official [documentation](https://docs.astral.sh/uv/reference/cli/#uv-venv)

````python
!cd $HOME/tmp/demo-project && uv venv --verbose
````

````python
!tree -a -L 5 $HOME/tmp/demo-project/.venv
````

The ``uv venv`` command is used to spawn a *PEP* conform **virtual environment** containing the common components. These components are:
* the ``bin`` folder: stores the activation scripts, the symbolic links to the python interpreter version and python launcher scripts for cli based applications *(will be discussed later)*
* the ``lib`` folder: contains any kind of third-party application that we are going to use within our project
* a ``pyvenv.cfg`` file: the basic configuration file of the virtual environment

However, the ``venv`` command is not installing any in the ``pyproject.toml`` referenced third-party libraries, this needs to be done with the other ``uv`` commands. 

---

To install a dependency we can use the ``uv add`` command as defined in the [documentation](https://docs.astral.sh/uv/reference/cli/#uv-add). A suitable test candidate is [click](https://click.palletsprojects.com/en/stable/) - this ios a single, standalone libary without any further dependencies.

We are deleting the ``.venv`` folder and installing the dependency afterwards.

````python
!rm -rf $HOME/tmp/demo-project/.venv && cd $HOME/tmp/demo-project && uv add click==1.0.0 --verbose
````

````python
!ls $HOME/tmp/demo-project && echo "---" && tree -a -L 4 $HOME/tmp/demo-project/.venv && echo "---" && cat $HOME/tmp/demo-project/pyproject.toml
````

##### Conclusion

To sum it up, the ``uv add`` command does the following for us:
1. it creates a virtual environment
2. it installs the required dependencies insie the ``site-packages`` folder
3. it keeps the ``pyproject.toml`` up to date by adding the dependency inside the `dependencies[]` array
4. it create a  ``uv.lock`` file *(we are talking about the file later on)*

#### 1.4.2 Setup Python 3.9

There are multiple ways how to setup Python on a running system. One option is to install it via an ppa remote. It supports the installation of multiple Python versions side-by-side on the system. However the installation at least requires some basic system knowledge. 

The official [instructions](https://launchpad.net/~deadsnakes/+archive/ubuntu/ppa) are offering the following guide

```

Display sources.list entries for: 
Noble (24.04)
deb https://ppa.launchpadcontent.net/deadsnakes/ppa/ubuntu noble main 
deb-src https://ppa.launchpadcontent.net/deadsnakes/ppa/ubuntu noble main 
Signing key:
4096R/F23C5A6CF475977595C89F51BA6932366A755776 (What is this?)
Fingerprint:

```

```bash
# get the signing key and convert it in apt's keyring binary format
curl -fsSL "https://keyserver.ubuntu.com/pks/lookup?op=get&search=0xF23C5A6CF475977595C89F51BA6932366A755776" | gpg --dearmor | sudo tee /etc/apt/keyrings/deadsnakes.gpg >/dev/null
```

```bash
# setup the remotes
cat <<EOF | sudo tee /etc/apt/sources.list.d/deadsnakes.list
deb [trusted=yes signed-by=/etc/apt/keyrings/deadsnakes.gpg] https://ppa.launchpadcontent.net/deadsnakes/ppa/ubuntu noble main
deb-src [trusted=yes signed-by=/etc/apt/keyrings/deadsnakes.gpg] https://ppa.launchpadcontent.net/deadsnakes/ppa/ubuntu noble main
EOF
```


#### 1.4.2 Managing python interpreter versions with uv

Take a look above and you can see that the ``.venv`` folder actually contains a *bin/python3* with a symbolic link to ``/usr/bin/python3.12`` - our system wide installed python.

In contrast to other package managers, ``uv`` is also able to manage the lifecycle of *Python* itself on our system. To manage the ``Python`` version of our project, ``uv`` provides the ```uv python``` [command](https://docs.astral.sh/uv/guides/install-python/), which offers some extended features like installation, uninstallation, pinning etc. in order to deal with the interpreter version. 

We are slightly modifying the ``main.py`` and insert a small section into it:

```python
import sys

print(sys.executable, sys.version)
```

This command prints the location and version info of the ``Python`` interpreter. With the ``uv run`` command we can directly invoke our project implementation without taking care of the virtual environment. This is a feature that uv takes care of for us. 

````python
!cd $HOME/tmp/bobs-server && export UV_NATIVE_TLS=1 && uv python install 3.9
````

## 3. Caching with uv

### 3.1 Excurse: Layer analyzation with Docker

One of ``uv``'s self mentioned benefits is It's caching which is optimized also across different virtual environments. According to the official [caching documentation](https://docs.astral.sh/uv/concepts/cache/) it comes with a series of advantages. However, often it is really hard to see how different commands are changing the operating system. To overcome this issue, we are using a common strategy which is named *layer analyzation. In particular we are faciliating the fact that oci images are built up on certain layers. These layers can be picked and analyzed and they give us a clear understanding about the system change. 

To understand the core idea we are using a very simple base image where we are going to [install](https://docs.astral.sh/uv/getting-started/installation/#standalone-installer) ``uv``.

```Dockerfile
FROM python:3.12-slim

RUN apt-get update \
    && apt-get install -y --no-install-recommends curl \
    && rm -rf /var/lib/apt/lists/*

RUN curl -LsSf https://astral.sh/uv/0.11.28/install.sh -o /tmp/install.sh \
    && chmod +x /tmp/install.sh

RUN /tmp/install.sh
```

Once we have create the file we are going to do the following:
* building the image
* converting the image into a file system container
* picking the last layer
* extracting the last layer

````python
# defining some variablesW
TARGET_DIR="$HOME/tmp/cache_analysis"
````

````python
# copying the image into a temp folder
!mkdir -p $TARGET_DIR 

# copy the Dockerfile into the temp folder
!cp Dockerfile $TARGET_DIR/Dockerfile

# build the image using docker
!cd $TARGET_DIR && DOCKER_BUILDKIT=1 docker build -q -t "uv-base-image:1.0.0" -f Dockerfile .
````

````python
TARGET_DIR="$HOME/tmp/cache_analysis"
# save the image as a file system container
!docker save "$IMAGE_TAG" -o "$TARGET_DIR/image.tar" && ls $TARGET_DIR
````

````python
# extracting the image 
!TARGET_DIR="$HOME/tmp/cache_analysis"; mkdir -p "$TARGET_DIR/image" && tar -xf "$TARGET_DIR/image.tar" -C "$TARGET_DIR/image" && ls -la $TARGET_DIR/image
````

````python
# checking the image manifest
!MANIFEST="$HOME/tmp/cache_analysis/image/manifest.json"; cat $MANIFEST
````

````python
!rm -rf "$HOME/tmp/cache_analysis/image/last_layer"
!mkdir -p "$HOME/tmp/cache_analysis/image/last_layer"

# extracting the last layer
!LAST_LAYER=$(jq -r '.[0].Layers[-1]' "$HOME/tmp/cache_analysis/image/manifest.json"); \
echo "$LAST_LAYER"; \
tar -xf "$HOME/tmp/cache_analysis/image/$LAST_LAYER" -C "$HOME/tmp/cache_analysis/image/last_layer"
````

````python
# taking a look
!tree -a $HOME/tmp/cache_analysis/image/last_layer
````

With the so called *Last Layer analyzation* we are able to visualize the system change of certain layer commands of our oci image build. Basically we are influencing the container build and simply picking the dedicated layer for analyzation. This approach is adaptable for a huge amount of various usecases. We are going to use this approach within our course to familiarze ourselves a bit more with the impact of certain commands.
