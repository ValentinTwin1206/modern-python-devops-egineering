# Virtual Environments

When working on a Python project it is very common to work in so called *Virtual Environments*, a lightweight feature of Python's standard lib that isolates the project libraries from the system-wide installed packages. 

Using virtual environments is beneficial for some reasons:
* isolation enables the work on multiple projects without raising dependency conflicts
* you can work with multiple python interpreters in parallel
* it improves the overview of the used packages
* it is easy to clean up without the fear to break the system

*Virtual environments* follow the **PEP-405** standard that describe the exact layout and components of a virtual environment.

### Work within an virtual environment

We have seen that a virtual environment follows a standard directory structure and includes its own Python interpreter in the ``bin`` directory. We can use this interpreter (instead of the system-wide Python) inside the virtual environment.

```python
# with the sys.executable, we can see the actual interpreter location
!$HOME/tmp/proj1_pyguard/.venv/bin/python3 -c "import sys; print('Python packages:', sys.executable, sys.version)"
```


Another common way to work with a virtual environment is to use the ``activate`` script. Running this script activates the virtual environment, allowing you to work within it without having to explicitly invoke the Python interpreter from the bin directory for every command.

```python
# the activate script
!cat $HOME/tmp/proj1_pyguard/.venv/bin/activate
```

The activate script sets the *VIRTUAL_ENV* and *VIRTUAL_ENV_PROMPT* environment variables and prepends the virtual environment's ``bin`` directory to the system ``PATH``. In our example, it adds ``$HOME/tmp/proj1_pyguard/.venv/bin`` to the ``PATH``, allowing you to invoke the executables in that directory directly without specifying their full paths.

### The ``bin`` folder

The *bin* directory is one of the key components of a virtual environment. It contains the activation scripts, the Python interpreter, and the `pip` command-line interface. More generally, it holds the executables, symlinks, and wrapper scripts associated with the virtual environment.

Many Python packages provide their own command-line interface (CLI), allowing users to invoke the application directly from the terminal. To demonstrate this, we will install our first dependency, `pytest`, a widely used testing framework for Python. In addition to its Python library, `pytest` also provides a CLI entry point that can be invoked directly from the command line.

```python
# we are going to install our first dependency inside the venv
!cd $HOME/tmp/proj1_pyguard && .venv/bin/python3 -m pip install pytest
```

```python
# what changed compared to the initial setup
!tree -a -L 4 $HOME/tmp/proj1_pyguard/.venv/
```

```python
# let's take a deeper look at the pytest executable
!cat $HOME/tmp/proj1_pyguard/.venv/bin/pytest
```

The script begins with a *shebang* that points to the Python interpreter inside the virtual environment, ensuring that `pytest` is executed with the correct interpreter and installed packages. The rest of the script imports `console_main()` from the `pytest` package and then calls `console_main()`.
This wrapper pattern is common in the Python ecosystem. Whenever a package provides a command-line interface, `pip` generates a small launcher script in the virtual environment's *bin* directory that delegates execution to the package's Python code.

### The ``lib`` folder

The ``lib`` folder on the other hand contains the third-party packages that are installed inside the virtual environment and which the interpreter is used. By default, the virtual environment only has access to the third-party packages defined inside the ``lib`` folder. This isolation enables multiple installation of the same package on the system without causing a dependency conflict.

```python
# using the pip interface lists the installed packages
!$HOME/tmp/proj1_pyguard/.venv/bin/pip3 freeze
```

```python
!$HOME/tmp/proj1_pyguard/.venv/bin/python3 -c "import sys; print('Python packages:', sys.path)"
```

### Summary

*Virtual environments* follow the **PEP-405** standard that describe the exact layout and components of a virtual environment. The most important components are:
1. the ``pyvenv.cfg`` - the central configuration file that tells Python that a directory is a virtual environment
2. the ``bin`` folder - this folder holds symlinks and wrapper binaries
3. the ``lib`` folder - this folder contains the third-party packages
4. the ``include`` folder - this folder contains c header files

### Editable dependencies

When you are developing a Python library locally, you usually want to test changes in your target project as quickly as possible. Packaging, publishing, downloading, and reinstalling the library after every small change would create a very slow development cycle.

To avoid that, Python allows us to install a library as an **editable dependency**. Instead of copying the package into the virtual environment, the package manager creates a link to the library's source code. This means every change you make in the library becomes immediately available in the target project without reinstalling it.

```python
# we are going to make an editable dependency installation
!cd $HOME/tmp/proj1_pyguard/ && $HOME/tmp/proj1_pyguard/.venv/bin/python3 -m pip install -e . 
```

```python
!$HOME/tmp/proj1_pyguard/.venv/bin/python3 -m pip list
```

```python
# what changed compared to the initial setup
!tree -a -L 4 $HOME/tmp/proj1_pyguard/.venv/
```

```python
!cat $HOME/tmp/proj1_pyguard/.venv/lib/python3.9/site-packages/__editable__.pyguard-0.1.0.pth
```

```python
!$HOME/tmp/proj1_pyguard/.venv/bin/python3 -c "import sys; print('Python package path:', sys.path)"
```
